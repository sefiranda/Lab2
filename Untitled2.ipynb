###przykład 1

```python
liczby = [10, 20, 30, 40]

suma = sum(liczby)
srednia = suma / len(liczby)

print("Liczby:", liczby)
print("Średnia:", srednia)
```

###przykład 2




```python
n = int(input("Podaj liczbę całkowitą: "))

if n % 2 == 0:
    print(n, "jest parzysta")
else:
    print(n, "jest nieparzysta")
```

```python
def przywitaj(imie):
    print("Cześć,", imie, "!")

imiona = ["Ania", "Bartek", "Kasia"]

for i in imiona:
    przywitaj(i)
```
